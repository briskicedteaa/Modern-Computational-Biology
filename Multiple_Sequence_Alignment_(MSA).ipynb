# Multiple Sequence Alignment with **Clustal** **Omega**

### MSA is a conputational process that takes three or more text strings representing DNA, RNA, nucleotides, or proteins, and aligns them vertically so that the maximum number of matching characters line up in columns.

### For this assigment, I chose 5 TrkB homologs from BLAST and aligned their sequences on the EMBL-EBI Clustal Omega web server.

### My **main objective** was to compare the sequences and identify conserved (high similar) and variable regiona (insertions or deletions).

###**Takeaway:** MSA extends sequence comparison beyond two proteins by aligning an entire family of homologous sequences at once. Unlike pairwise alignment, which provides an overall similarity score, MSA reveals conserved amino acids, variable regions, and insertion/deletion events across multiple proteins. This makes it a valuable tool for identifying functionally important residues, studying evolutionary relationships, and guiding downstream analyses such as mutation studies, structural biology, and protein engineering.
